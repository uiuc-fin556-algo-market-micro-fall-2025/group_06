```python
# --- Calculate Technical Indicators for 1hr Data ---
import pandas as pd
import ta

# Load the 1hr data
df_1hr = pd.read_csv('copper_continuous_1hr_jan2024.csv', index_col=0, parse_dates=True)

# Calculate SMA (Simple Moving Averages)
df_1hr['SMA_20'] = ta.trend.SMAIndicator(close=df_1hr['close'], window=20).sma_indicator()
df_1hr['SMA_50'] = ta.trend.SMAIndicator(close=df_1hr['close'], window=50).sma_indicator()
df_1hr['SMA_200'] = ta.trend.SMAIndicator(close=df_1hr['close'], window=200).sma_indicator()

# Calculate EMA (Exponential Moving Averages)
df_1hr['EMA_12'] = ta.trend.EMAIndicator(close=df_1hr['close'], window=12).ema_indicator()
df_1hr['EMA_26'] = ta.trend.EMAIndicator(close=df_1hr['close'], window=26).ema_indicator()
df_1hr['EMA_50'] = ta.trend.EMAIndicator(close=df_1hr['close'], window=50).ema_indicator()

# Calculate Bollinger Bands
bb_indicator = ta.volatility.BollingerBands(close=df_1hr['close'], window=20, window_dev=2)
df_1hr['BB_upper'] = bb_indicator.bollinger_hband()
df_1hr['BB_middle'] = bb_indicator.bollinger_mavg()
df_1hr['BB_lower'] = bb_indicator.bollinger_lband()

# Calculate RSI (Relative Strength Index)
df_1hr['RSI'] = ta.momentum.RSIIndicator(close=df_1hr['close'], window=14).rsi()

# Calculate MACD
macd_indicator = ta.trend.MACD(close=df_1hr['close'])
df_1hr['MACD'] = macd_indicator.macd()
df_1hr['MACD_signal'] = macd_indicator.macd_signal()
df_1hr['MACD_hist'] = macd_indicator.macd_diff()

# Calculate Volume indicators (simple moving average of volume)
df_1hr['Volume_SMA'] = df_1hr['volume'].rolling(window=20).mean()

print("Technical indicators calculated for 1hr data")
print(f"Columns: {list(df_1hr.columns)}")
```

```python
# --- Calculate Technical Indicators for 1min Data ---
# Load the 1min data
df_1m = pd.read_csv('copper_continuous_1min_jan2024.csv', index_col=0, parse_dates=True)

# Calculate SMA (Simple Moving Averages)
df_1m['SMA_20'] = ta.trend.SMAIndicator(close=df_1m['close'], window=20).sma_indicator()
df_1m['SMA_50'] = ta.trend.SMAIndicator(close=df_1m['close'], window=50).sma_indicator()
df_1m['SMA_200'] = ta.trend.SMAIndicator(close=df_1m['close'], window=200).sma_indicator()

# Calculate EMA (Exponential Moving Averages)
df_1m['EMA_12'] = ta.trend.EMAIndicator(close=df_1m['close'], window=12).ema_indicator()
df_1m['EMA_26'] = ta.trend.EMAIndicator(close=df_1m['close'], window=26).ema_indicator()
df_1m['EMA_50'] = ta.trend.EMAIndicator(close=df_1m['close'], window=50).ema_indicator()

# Calculate Bollinger Bands
bb_indicator = ta.volatility.BollingerBands(close=df_1m['close'], window=20, window_dev=2)
df_1m['BB_upper'] = bb_indicator.bollinger_hband()
df_1m['BB_middle'] = bb_indicator.bollinger_mavg()
df_1m['BB_lower'] = bb_indicator.bollinger_lband()

# Calculate RSI (Relative Strength Index)
df_1m['RSI'] = ta.momentum.RSIIndicator(close=df_1m['close'], window=14).rsi()

# Calculate MACD
macd_indicator = ta.trend.MACD(close=df_1m['close'])
df_1m['MACD'] = macd_indicator.macd()
df_1m['MACD_signal'] = macd_indicator.macd_signal()
df_1m['MACD_hist'] = macd_indicator.macd_diff()

# Calculate Volume indicators (simple moving average of volume)
df_1m['Volume_SMA'] = df_1m['volume'].rolling(window=20).mean()

print("Technical indicators calculated for 1min data")
print(f"Columns: {list(df_1m.columns)}")
```

```python
# --- Visualize 1hr Data with Technical Indicators ---
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Create subplots: Price with indicators, RSI, MACD, Volume
fig = make_subplots(
    rows=4, cols=1,
    shared_xaxes=True,
    vertical_spacing=0.05,
    subplot_titles=('Copper Futures - 1hr OHLC with Indicators', 'RSI', 'MACD', 'Volume'),
    row_heights=[0.4, 0.15, 0.15, 0.3],
    specs=[[{"secondary_y": False}],
           [{"secondary_y": False}],
           [{"secondary_y": False}],
           [{"secondary_y": False}]]
)

# Add candlestick chart
fig.add_trace(
    go.Candlestick(
        x=df_1hr.index,
        open=df_1hr['open'],
        high=df_1hr['high'],
        low=df_1hr['low'],
        close=df_1hr['close'],
        name='Price'
    ),
    row=1, col=1
)

# Add SMA lines
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['SMA_20'], name='SMA 20', line=dict(color='blue', width=1)),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['SMA_50'], name='SMA 50', line=dict(color='orange', width=1)),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['SMA_200'], name='SMA 200', line=dict(color='red', width=1)),
    row=1, col=1
)

# Add EMA lines
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['EMA_12'], name='EMA 12', line=dict(color='purple', width=1, dash='dash')),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['EMA_26'], name='EMA 26', line=dict(color='cyan', width=1, dash='dash')),
    row=1, col=1
)

# Add Bollinger Bands
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['BB_upper'], name='BB Upper', 
               line=dict(color='rgba(128,128,128,0.5)', width=1), showlegend=False),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['BB_middle'], name='BB Middle', 
               line=dict(color='rgba(128,128,128,0.5)', width=1), showlegend=False),
    row=1, col=1
)
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['BB_lower'], name='BB Lower', 
               line=dict(color='rgba(128,128,128,0.5)', width=1), 
               fill='tonexty', fillcolor='rgba(128,128,128,0.1)', showlegend=True),
    row=1, col=1
)

# Add RSI
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['RSI'], name='RSI', line=dict(color='purple', width=2)),
    row=2, col=1
)
# Add RSI overbought/oversold lines
fig.add_hline(y=70, line_dash="dash", line_color="red", opacity=0.5, row=2, col=1)
fig.add_hline(y=30, line_dash="dash", line_color="green", opacity=0.5, row=2, col=1)

# Add MACD
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['MACD'], name='MACD', line=dict(color='blue', width=2)),
    row=3, col=1
)
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['MACD_signal'], name='MACD Signal', line=dict(color='red', width=2)),
    row=3, col=1
)
# Add MACD histogram
colors_macd = ['green' if val >= 0 else 'red' for val in df_1hr['MACD_hist']]
fig.add_trace(
    go.Bar(x=df_1hr.index, y=df_1hr['MACD_hist'], name='MACD Hist', marker_color=colors_macd),
    row=3, col=1
)

# Add volume bars
colors = ['red' if df_1hr['close'].iloc[i] < df_1hr['open'].iloc[i] else 'green' 
          for i in range(len(df_1hr))]
fig.add_trace(
    go.Bar(x=df_1hr.index, y=df_1hr['volume'], name='Volume', marker_color=colors),
    row=4, col=1
)
# Add volume SMA
fig.add_trace(
    go.Scatter(x=df_1hr.index, y=df_1hr['Volume_SMA'], name='Volume SMA', 
               line=dict(color='orange', width=2)),
    row=4, col=1
)

# Update layout
fig.update_layout(
    title='Copper Futures Continuous Contract (HG.v.0) - 1 Hour with Technical Indicators - January 2024',
    height=1200,
    showlegend=True,
    xaxis_rangeslider_visible=False,
    hovermode='x unified'
)

# Update y-axis labels
fig.update_yaxes(title_text="Price ($)", row=1, col=1)
fig.update_yaxes(title_text="RSI", row=2, col=1, range=[0, 100])
fig.update_yaxes(title_text="MACD", row=3, col=1)
fig.update_yaxes(title_text="Volume", row=4, col=1)

# Show the plot
fig.show()
print(f"Total data points: {len(df_1hr)}")
print(f"Date range: {df_1hr.index.min()} to {df_1hr.index.max()}")
```
